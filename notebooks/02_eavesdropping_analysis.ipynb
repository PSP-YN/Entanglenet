```
from src.experiments import run_bb84
import matplotlib.pyplot as plt

eve_levels = [0.0, 0.10, 0.25, 0.50, 0.75, 1.0]
qbers = []
for e in eve_levels:
    res = run_bb84(1000, e, 0.0, seed=42)
    qbers.append(res['qber'])

plt.plot(eve_levels, qbers, marker='o')
plt.xlabel('Eve Probability')
plt.ylabel('QBER')
plt.title('QBER vs Eve')
plt.show()
```
